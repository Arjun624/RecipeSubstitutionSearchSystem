# Recipe Substitution Search System Demo

This notebook demonstrates the Recipe Substitution Search System functionality.

```python
import os
import sys

# Add parent directory to path
sys.path.append(os.path.abspath('..'))

from src.data.recipe_dataset import RecipeDataset
from src.data.substitution_db import SubstitutionDatabase
from src.models.search_engine import RecipeSearchEngine, SubstitutionSearchEngine
```

```python
# Load data
dataset = RecipeDataset()
substitution_db = SubstitutionDatabase()

# Create search engines
basic_engine = RecipeSearchEngine(dataset, substitution_db)
enhanced_engine = SubstitutionSearchEngine(dataset, substitution_db)

print(f"Loaded {len(dataset)} recipes with {len(dataset.annotations)} annotated queries")
```

```python
# Try a search query
query = "vegetarian pad thai"
results = enhanced_engine.search(query, top_k=5)

print(f"Search results for "{query}":")for i, (recipe, score) in enumerate(results):
    print(f"
Result #{i+1}: {recipe.title} (Score: {score:.2f})")
    print(f"Diet labels: {recipe.diet_labels}")
    print(f"Ingredients: {len(recipe.ingredients)} items")
```
